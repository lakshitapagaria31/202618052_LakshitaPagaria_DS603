## Problem 1: Largest, Smallest, Sum and Average of a List

```python
## Write a Python program to accept 10 integers from the user, store them in a list, and display the largest element, the smallest element, the sum of all elements, and the average of the list.

numbers = list(map(int, input().split()))

largest = max(numbers)
smallest = min(numbers)
total_sum = sum(numbers)
average = total_sum / len(numbers)

print(f"Largest: {largest}")
print(f"Smallest: {smallest}")
print(f"Sum: {total_sum}")
print(f"Average: {average:.2f}")
```

## Problem 2: Days of the Week Using a Tuple


```python
## Write a Python program to create a tuple containing the names of the seven days of the week. Display the first three days using slicing, then unpack the tuple into separate variables and print each day.

days = ("Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday")

first_three = days[:3]
print("First Three Days:", " ".join(first_three))

mon, tue, wed, thu, fri, sat, sun = days
print(mon)
print(tue)
print(wed)
print(thu)
print(fri)
print(sat)
print(sun)
```

## Problem 3: Employee Details Using a Dictionary

```python
## Write a Python program to create a dictionary that stores the details of an employee: Employee ID, Name, Department, and Salary. Update the employee's salary, add the employee's email address, and display all key-value pairs.

emp_id = input()
name = input()
dept = input()
initial_salary = int(input())
updated_salary = int(input())
email = input()

employee = {
    "EmployeeID": emp_id,
    "Name": name,
    "Department": dept,
    "Salary": initial_salary
}

employee["Salary"] = updated_salary
employee["Email"] = email

for key, value in employee.items():
    print(f"{key}: {value}")
```

## Problem 4: Set Operations and Disjoint Check

```python
## Write a Python program to create two sets of integers entered by the user. Display their union, intersection, difference, and symmetric difference. Also determine whether the two sets are disjoint.

n1 = int(input())
set1 = set(map(int, input().split()))

n2 = int(input())
set2 = set(map(int, input().split()))

union_set = set1 | set2
intersection_set = set1 & set2
difference_set = set1 - set2
symmetric_difference_set = set1 ^ set2
is_disjoint = "Yes" if set1.isdisjoint(set2) else "No"

print(f"Union: {union_set}")
print(f"Intersection: {intersection_set}")
print(f"Difference: {difference_set}")
print(f"Symmetric Difference: {symmetric_difference_set}")
print(f"Disjoint: {is_disjoint}")
```

## Problem 5: Remove Duplicate Elements Using a Set

```python
## Write a Python program to remove duplicate elements from a list using a set. Display the original list and the list after removing duplicates, preserving the order of first occurrence.

n = int(input())
original_list = list(map(int, input().split()))

seen = set()
deduplicated_list = []

for item in original_list:
    if item not in seen:
        seen.add(item)
        deduplicated_list.append(item)

print(f"Original List: {original_list}")
print(f"List After Removing Duplicates: {deduplicated_list}")
```

## Problem 6: Marks of Five Students Using a Dictionary

```python
## Write a Python program to maintain the marks of five students using a dictionary, where the student's name is the key and the marks are the value. Display all students with their marks, calculate the average marks, and identify the student with the highest marks.

students = {}

for _ in range(5):
    line = input().split()
    name = line[0]
    marks = int(line[1])
    students[name] = marks

total_marks = 0
highest_marks = -1
highest_scorer = ""

for name, marks in students.items():
    print(f"{name}: {marks}")
    total_marks += marks
    if marks > highest_marks:
        highest_marks = marks
        highest_scorer = name

average = total_marks / 5
print(f"Average Marks: {average:.2f}")
print(f"Highest Marks: {highest_scorer} ({highest_marks})")
```
