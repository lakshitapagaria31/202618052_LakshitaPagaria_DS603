# Problem 1: Valid Palindrome

## Design Approach

- Use the **two-pointer approach** with one pointer starting from the left and the other from the right.
- Skip all non-alphanumeric characters from both sides.
- Compare the characters after converting them to lowercase.
- If any pair of characters is different, the string is not a palindrome.
- Continue moving both pointers toward the center.
- If all valid characters match, return `true`.

## Complexity

- **Time Complexity:** `O(n)`
- **Space Complexity:** `O(1)`

```python
s = input()

left = 0
right = len(s) - 1

is_palindrome = True

while left < right:
    while left < right and not s[left].isalnum():
        left += 1

    while left < right and not s[right].isalnum():
        right -= 1

    if s[left].lower() != s[right].lower():
        is_palindrome = False
        break

    left += 1
    right -= 1

print("true" if is_palindrome else "false")
```

# Problem 2: Valid Parentheses

## Design Approach

- Use a **stack** to store opening brackets.
- When an opening bracket `(`, `{`, or `[` is encountered, push it onto the stack.
- When a closing bracket is encountered, check whether the top of the stack contains its corresponding opening bracket.
- If it does, remove the opening bracket using `pop()`.
- If it does not, the parentheses are invalid.
- At the end, the stack must be empty for the string to be valid.

## Complexity

- **Time Complexity:** `O(n)`
- **Space Complexity:** `O(n)`

```python
s = input()

stack = []
pairs = {
    ')': '(',
    '}': '{',
    ']': '['
}

is_valid = True

for char in s:
    if char in '({[':
        stack.append(char)
    else:
        if not stack or stack[-1] != pairs[char]:
            is_valid = False
            break
        stack.pop()

if stack:
    is_valid = False

print("true" if is_valid else "false")
```

# Problem 3: Min Stack

## Design Approach

- Use **two stacks**:
  - `stack` stores all elements.
  - `min_stack` stores the minimum elements.
- During `PUSH`, insert the value into the main stack.
- If the value is smaller than or equal to the current minimum, also push it into `min_stack`.
- During `POP`, if the removed value is the current minimum, remove it from `min_stack`.
- `TOP` returns the top element of the main stack.
- `GETMIN` returns the top element of `min_stack`.
- This allows the minimum element to be retrieved in constant time.

## Complexity

- **PUSH:** `O(1)`
- **POP:** `O(1)`
- **TOP:** `O(1)`
- **GETMIN:** `O(1)`
- **Overall Time Complexity:** `O(Q)`
- **Space Complexity:** `O(n)`

```python
class MinStack:
    def __init__(self):
        self.stack = []
        self.min_stack = []

    def push(self, value):
        self.stack.append(value)

        if not self.min_stack or value <= self.min_stack[-1]:
            self.min_stack.append(value)

    def pop(self):
        value = self.stack.pop()

        if value == self.min_stack[-1]:
            self.min_stack.pop()

    def top(self):
        return self.stack[-1]

    def getMin(self):
        return self.min_stack[-1]


Q = int(input())

min_stack = None

for _ in range(Q):
    operation = input().split()

    if operation[0] == "MinStack":
        min_stack = MinStack()

    elif operation[0] == "PUSH":
        min_stack.push(int(operation[1]))

    elif operation[0] == "POP":
        min_stack.pop()

    elif operation[0] == "TOP":
        print(min_stack.top())

    elif operation[0] == "GETMIN":
        print(min_stack.getMin())
```

# Problem 4: Implement Queue using Stacks

## Design Approach

- Use **two stacks**, `stack1` and `stack2`.
- `stack1` is used to insert new elements.
- `stack2` is used to remove or access elements from the front.
- For `PUSH`, insert the element into `stack1`.
- For `POP` or `PEEK`, if `stack2` is empty, transfer all elements from `stack1` to `stack2`.
- This reversal makes the oldest element available at the top of `stack2`.
- `POP` removes the top element of `stack2`.
- `PEEK` returns the top element without removing it.
- The queue is empty only when both stacks are empty.

## Complexity

- **PUSH:** `O(1)`
- **POP:** `O(1)` amortized, `O(n)` worst case
- **PEEK:** `O(1)` amortized, `O(n)` worst case
- **EMPTY:** `O(1)`
- **Overall Time Complexity:** `O(Q)` amortized
- **Space Complexity:** `O(n)`

```python
class MyQueue:
    def __init__(self):
        self.stack1 = []
        self.stack2 = []

    def push(self, x):
        self.stack1.append(x)

    def pop(self):
        self._transfer()
        return self.stack2.pop()

    def peek(self):
        self._transfer()
        return self.stack2[-1]

    def empty(self):
        return not self.stack1 and not self.stack2

    def _transfer(self):
        if not self.stack2:
            while self.stack1:
                self.stack2.append(self.stack1.pop())


Q = int(input())

queue = None

for _ in range(Q):
    operation = input().split()

    if operation[0] == "MyQueue":
        queue = MyQueue()

    elif operation[0] == "PUSH":
        queue.push(int(operation[1]))

    elif operation[0] == "POP":
        print(queue.pop())

    elif operation[0] == "PEEK":
        print(queue.peek())

    elif operation[0] == "EMPTY":
        print("true" if queue.empty() else "false")
```

# Problem 5: Length of Last Word

## Design Approach

- Start scanning the string from the **right side**.
- First skip any trailing spaces.
- Then continue moving left while the characters are not spaces.
- Count these characters.
- The count represents the length of the last word.

## Complexity

- **Time Complexity:** `O(n)`
- **Space Complexity:** `O(1)`

```python
s = input()

i = len(s) - 1

while i >= 0 and s[i] == ' ':
    i -= 1

length = 0

while i >= 0 and s[i] != ' ':
    length += 1
    i -= 1

print(length)
```

# Problem 6: Longest Substring Without Repeating Characters

## Design Approach

- Use the **sliding window technique**.
- Maintain a window from `left` to `right` containing characters without duplicates.
- Store the latest index of each character in `last_seen`.
- Move `right` through the string one character at a time.
- If the current character has already appeared inside the current window, move `left` to one position after its previous occurrence.
- Calculate the current window length and update `max_length`.
- Continue until the entire string has been processed.

## Complexity

- **Time Complexity:** `O(n)`
- **Space Complexity:** `O(n)`

```python
s = input()

left = 0
max_length = 0
last_seen = {}

for right in range(len(s)):
    if s[right] in last_seen and last_seen[s[right]] >= left:
        left = last_seen[s[right]] + 1

    last_seen[s[right]] = right

    max_length = max(max_length, right - left + 1)

print(max_length)
```
