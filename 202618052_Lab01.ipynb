### 1. Print the following pattern:
- 1
- 2 2
- 3 3 3
- 4 4 4 4

```python
n = int(input("Enter a number: "))
for i in range(1, n+1):
    for j in range(1, i+1):
        print(i, end = ' ')
    print()
```

### 2. Write a program to print the first N Fibonacci numbers, where N is user-defined.

- 0, 1, 1, 2, 3, 5, 8, 13, 21, ...


```python
n = int(input("Enter a number upto which you need to print: "))
a = 0
b = 1
for i in range(n):
    print(a, end = ' ')
    c = a + b
    a = b 
    b = c
```

### 3. Print all prime numbers between 100 and 1000.


```python
for i in range(100, 1001):
    for j in range(2, int(i**0.5) + 1):
        if i % j == 0:
            break
    else:
        print(i, end = ' ')
```

### 4. Write a Python program to calculate the Simple Interest (SI) and the Total Amount after interest is added. The program should accept the Principal Amount (P), Rate of Interest (R) (in percentage per annum), and Time Period (T) (in years) as input from the user. Calculate the Simple Interest using the formula SI = (P × R × T) / 100, then calculate the Total Amount as Principal + Simple Interest.Finally, display both values in a clear and readable format.

```python
P = float(input("Enter the principal amount: "))
R = float(input("Enter the rate of interest: "))
T = float(input("Enter the time period: "))

## SI Calculate 
simple_interest = (P*R*T)/100
total_amount = P + simple_interest

## Printing the results
print(f"Simple Interest: {simple_interest:.4f}")
print(f"Total Amount: {total_amount:.4f}")

```

### 5. Find the largest among three numbers.


```python
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))
c = float(input("Enter third number: "))
if a >= b and a >= c:
    print(f"{a} is the largest")
elif b >= c: 
    print(f"{b} is the largest")
else:
    print(f"{c} is the largest")
```
